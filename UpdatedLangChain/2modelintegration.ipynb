```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:llama-3.3-70b-versatile")
response = model.invoke("You know whi is parrot?")
response
```

```python
## Streaming

model.stream("Hello, WHo is mayank Goayl, he studied in Gautam Buddha University.")

for chunks in model.stream("Hello, WHo is mayank Goayl, he studied in Gautam Buddha University."):
    print(chunks.text, end='|', flush=True)
```
