# Structured tool

```python
from langchain_community.tools import tool, StructuredTool
from pydantic import BaseModel, Field
```

```python
class schema(BaseModel):
  a : int = Field(description="First Number")
  b : int = Field(description="Second Number")
```

```python
def multi_ply(a:int, b:int) -> int:
  return a*b
```

```python
multipli_fun = StructuredTool.from_function(
    name="multi_ply",
    description="Multiply two numbers",
    func=multi_ply,
    args_schema=schema
)
```

```python
multipli_fun.invoke({"a":3, "b":4})
```

# BAse Tool

```python
from langchain_community.tools import BaseTool
from pydantic import BaseModel, Field
from typing import Type
```

```python
class schema(BaseModel):
  a : int = Field(description="First Number")
  b : int = Field(description="Second Number")
```

```python
class Multipytool(BaseTool):
  name : str = "Multiply"
  description : str = "Multiply two numbers"

  args_schema: Type[BaseModel] = schema

  def _run(self, a:int, b:int) -> int:
    return a*b
```

```python
multiply_tool = Multipytool()
```

```python
result = multiply_tool.invoke({'a':3, 'b':3})

print(result)
print(multiply_tool.name)
print(multiply_tool.description)

print(multiply_tool.args)
```

# toolkit

```python
from langchain_core.tools import tool

# Custom tools
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
